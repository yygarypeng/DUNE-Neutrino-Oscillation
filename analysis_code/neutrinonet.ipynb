## *Oscillating neutrinos*
There are three flavors of neutrinos in the standard model, each corresponding to a lepton: electrons, muons, and tau leptons. They are, by definition massless particles. However,newer theories proposed otherwise, that neutrinos do have mass. The neutrino spinors are said to be **flavor eigenstates**, superpositions of mass eigenstates (solutions to the free Lagrangian)of different generations. If the neutrino generations are not identical in mass, the time evoluion would also be different, possibly resulting in a different flavor-eigenstate. This is described as the oscillation of neutrinos. Long baseline experiments such as DUNE measure the neutrinos from the source, where the neutrinos are in a coherent superposition, and at a far site, where the distance is sufficiently long for the phase shift to be measurable.

### *Quantifying the oscillation*
### [Pontecorvo–Maki–Nakagawa–Sakata (PMNS) matrix](https://en.wikipedia.org/wiki/Pontecorvo%E2%80%93Maki%E2%80%93Nakagawa%E2%80%93Sakata_matrix)  

#### Flavor mixing
$$
\begin{pmatrix} \nu_{e} \\ \nu_{\mu} \\ \nu_{\tau} \end{pmatrix} = 
\begin{pmatrix}
U_{e1} & U_{e2} & U_{e3} \\
U_{\mu 1} & U_{\mu 2} & U_{\mu 3} \\
U_{\tau 1} & U_{\tau 2} & U_{\tau 3}
\end{pmatrix}
\begin{pmatrix} \nu_{1} \\ \nu_{2} \\ \nu_{3} \end{pmatrix}
$$

#### Wolfenstein Parametrization
$c_{ij} \equiv \cos(\theta_{ij}) \qquad s_{ij} \equiv \sin(\theta_{ij})$
$$
\begin{pmatrix}
1 & 0 & 0 \\
0 & c_{23} & s_{23}\\
0 & -s_{23} & c_{23}\\
\end{pmatrix}
\begin{pmatrix}
 c_{13} & 0 & s_{13}e^{-i\delta_{CP}}\\
0 & 1 & 0 \\
-s_{13}e^{-i\delta_{CP}} & 0 c_{13} \\
\end{pmatrix}
\begin{pmatrix}
 c_{12} & s_{12} & 0\\
-s_{12} & c_{12} & 0\\
0 & 0 & 1 \\
\end{pmatrix}
$$


$$
=
\begin{pmatrix}
 c_{12}c_{13} & s_{12}c_{13} & s_{13}e^{-i\delta_{CP}}\\
-s_{12}c_{23} - c_{12}s_{23}s_{13}e^{i\delta_{CP}} &  c_{12}c_{13} - s_{12}s_{23}s_{13}e^{i\delta_{CP}} & s_{23}c_{13} \\
 s_{12}s_{23} - c_{12}s_{23}s_{13}e^{i\delta_{CP}} & -c_{12}c_{13} - s_{12}s_{23}s_{13}e^{i\delta_{CP}} & s_{23}c_{13} \\
\end{pmatrix}
$$


### How is it measured? 

#### Appearance Experiments:
Experiments measure transitions between different neutrino flavors. If the final flavor to be searched for in the detector is not present in the initial beam, the background can be very small. In this case, an experiment can be sensitive to rather small values of the imxing angle.
#### Disappearance Experiments:
By comparing the energy spectrum of a beam of neutrinos at the beam source and a far detector, this type of experiments measure the survival probability of a neutrino by counting the number of interactions in the detector and comparing it with the expected one. since, even in the absence of oscillations, the number of detected events have statistical fluctuations, it is very difficult to reveal a small disapperance. Therefore, in this type of expreiment, it is hard to measure small values of the mixing angle.

The signal for appearance is an excess of CC interactions over the expected background in the far detector.

### CP violation in the Standard Model
For particles $a$ and $b$, and respective antiparticles $\bar{a}$, $\bar{b}$  
Consider the processes $a \to b$ and $\bar{a} \to \bar{b}$ and corresponding amplitudes $\mathcal{M}$ and $\bar{\mathcal{M}}$.  
Without CP violation, the amplitudes are the same complex number. If a phase $\delta$ is picked up from the PMNS matrix, the amplitudes becomes:
* $\mathcal{M} = |\mathcal{M}| e^{i\theta}e^{+i\delta}$  
* $\bar{\mathcal{M}} = |\bar{\mathcal{M}}| e^{i\theta}e^{-i\delta}$  
The physically measureable amplitude $|\mathcal{M}|$ remains the same, however is the process has two different routes $a \to(1,2)\to b$  
* $\mathcal{M} = |\mathcal{M}_1| e^{i\theta_1}e^{+i\delta_1} + |\mathcal{M}_2| e^{i\theta_2}e^{+i\delta_2}$  
* $\bar{\mathcal{M}} = |\bar{\mathcal{M}}_1| e^{i\theta_1}e^{-i\delta_1} + |\bar{\mathcal{M}}_2| e^{i\theta_2}e^{-i\delta_2}$  


#### Transition probabilities: $P(\nu_a \to \nu_b) = P_\text{CP even}(\nu_a \to \nu_b) + P_\text{CP odd}(\nu_a \to \nu_b)$
Ref 3. [Neutrino mixing and CP-violation](https://cds.cern.ch/record/398912/files/9909254.pdf)
$$
P_\text{CP even}(\nu_\mu \to \nu_e) = \sin^2(\theta_{13})\sin^2(2\theta_{13})\sin^2 \left(\frac{\Delta{m_{23}}^2 L}{4E}\right)
$$
  
$$
P_\text{CP even}(\nu_e \to \nu_\tau) = \cos^2(\theta_{23})\sin^2(2\theta_{13})\sin^2 \left(\frac{\Delta{m_{23}}^2 L}{4E}\right)
$$
  
$$
P_\text{CP even}(\nu_\mu \to \nu_\tau) = \cos^4(\theta_{13})\sin^2(2\theta_{23})\sin^2 \left(\frac{\Delta{m_{23}}^2 L}{4E}\right)
$$

$$
P_\text{CP odd}(\nu_e \to \nu_\mu) = -P_\text{CP odd}(\nu_\mu \to \nu_\tau) = -P_\text{CP odd}(\nu_e \to \nu_\tau)
= -8\cos(\theta_{12})\cos^2(\theta_{13})\sin(\theta_{12})\sin(\theta_{13})\sin(\theta_{23})
\sin\left( \delta \frac{\Delta{m_{13}}^2 L}{4E}\right)\sin^2\left( \frac{\Delta{m_{23}}^2 L}{4E}\right)
$$

The entire complement of neutrino experiments to date has measured **five** of the mixing parameters: ($\theta_{12}.\theta_{23}.\theta_{13}$), and the two mass differences. $\Delta{m_{21}}^2$, $\Delta{m_{31}}^2$.  
However, only the difference is known, not the ordering. By convention, the index is chosen such that $\Delta{m_{21}}^2$ ("solar" mass splitting) is positive.  
The mass difference in the mass eigenstates: $\Delta{m_{31}}^2 = \Delta{m_{32}}^2 + \Delta{m_{21}}^2$   
* normal ordering: $m_1 < m_2 < m_3$
$$
m_1\underbrace{........}_{\Delta{m_{sol}}^2}m_2\underbrace{................}_{\Delta{m_{atm}}^2}m_3
$$
* Inverted ordering: $m_3 < m_1 < m_2$
$$
m_3\underbrace{...............}_{\Delta{m_{atm}}^2}m_1\underbrace{.......}_{\Delta{m_{sol}}^2}m_2
$$

### CP asymmetry
Consider the CP-asymmetry:
$$
A^{CP}_{a b} \equiv 
\frac{P(\nu_a \to \nu_b) - P(\bar{\nu}_a \to \bar{\nu}_b)}{P(\nu_a \to \nu_b) + P(\nu_a \to \nu_b)}
$$
This is a CP-odd observable. In a muon $\mu^+$ decay process $\mu^+ \to e^+ \nu_e \nu_\mu$, as the neutrino oscillates $\nu_e \to \nu_\mu$ , the $\nu_\mu$ charged current interactions produce negative muons $\mu^-$, also termed as a *wrong-sign* muon.
  
$P(\nu_\mu \to \nu_e)$ (and the conjugate) is estimated by measurements of *wrong-sign* muons.  
In practice, this observable is constructed from the integrated neutrino spectrum of decaying leptons $N_{\ell^\pm}$ and *wrong-sign* leptons $N^\text{wrong}_{\ell^\pm}$
$$
A_{e \mu}^\text{CP}(\delta) = \frac{N_{\mu^-}/N^\text{wrong}_{e^-} -  N_{\mu^+}/N^\text{wrong}_{e^+}}{ N_{\mu^-}/N^\text{wrong}_{e^-} +  N_{\mu^+}/N^\text{wrong}_{e^+}}
$$
This is a function of the CP phase $\delta$  

This new observable should demonstrate a clear relation between the training data point and the correct $\delta_{CP}$

```python
X_train_delta_e  = (np.sum(X_train[:,:,0],axis = 1) - np.sum(X_train[:,:,1],axis = 1))/(np.sum(X_train[:,:,0],axis = 1) + np.sum(X_train[:,:,1],axis = 1))
X_train_delta_mu = (np.sum(X_train[:,:,2],axis = 1) - np.sum(X_train[:,:,3],axis = 1))/(np.sum(X_train[:,:,2],axis = 1) + np.sum(X_train[:,:,3],axis = 1))
```

Again, we observe a clear separation between the *normal* and *inverted ordered* data points. This is more obvious in the $\nu_e$ appearance channel than compared to the $\nu_\mu$ disappearance channel

### general analysis codes

```python
from sklearn.metrics import mean_squared_error


def analysis(x_train_pred, y_train) -> None:
    # set statistical parameters
    n = len(x_train_pred)
    dof = n - 2
    t = scipy.stats.t.ppf(0.975, dof)

    # linear fitting line
    slope, intercept = np.polyfit(x_train_pred, y_train, 1)
    slope, intercept, r, p, se = scipy.stats.linregress(x_train_pred, y_train)
    plt_min = np.min([np.min(x_train_pred), np.min(y_train)])
    plt_max = np.max([np.max(x_train_pred), np.max(y_train)])
    x_line = np.linspace(plt_min, plt_max, 100)
    y_line = np.polyval([slope, intercept], x_line)

    # acceleration kernel (to calculate mse, confident level, and prediction level)
    @njit(fastmath=True)
    def mse_kernel(pred, true, t_value=t, n=n, dof=dof, x_line=x_line, y_line=y_line):
        if true.shape != pred.shape:
            raise ValueError("True/Pred data should be the same shape!")
        mse = np.empty(len(true))
        for ind, (t, p) in enumerate(zip(true, pred)):
            mse[ind] = np.average(np.square(t - p))
        pred_mean = np.average(x_train_pred)
        sum_se = np.sum(np.square(true - pred))
        std_error = np.sqrt(1 / dof * sum_se)
        # confidence interval
        ci = (
            t_value
            * std_error
            * np.sqrt(
                1 / n
                + np.square(x_line - pred_mean)
                / np.sum(np.square(x_train_pred - pred_mean))
            )
        )
        # predicting interval
        pi = (
            t_value
            * std_error
            * np.sqrt(
                1
                + 1 / n
                + np.square(x_line - pred_mean)
                / np.sum(np.square(x_train_pred - pred_mean))
            )
        )
        return mse, ci, pi

    mse, ci, pi = mse_kernel(x_train_pred, y_train)

    fig = plt.figure(figsize=(7, 7), dpi=80)
    plt.scatter(x_train_pred, y_train, s=1, alpha=0.05, label="data")
    plt.plot(x_line, y_line, "--", lw=2, c="red", label="Fitting")
    plt.plot(
        [plt_min, plt_max], [plt_min, plt_max], "-", lw=1, c="orange", label="theo line"
    )
    plt.plot(
        x_line, y_line + pi, "--", lw=1, c="tab:red", label="95% prediction interval"
    )
    plt.plot(x_line, y_line - pi, "--", lw=1, c="tab:red")
    # plt.plot(x_line, y_line + ci, '--', c='tab:grey', label='95% confidence interval')
    # plt.plot(x_line, y_line - ci, '--', c='tab:grey')
    plt.xlim(plt_min, plt_max)
    plt.ylim(plt_min, plt_max)
    plt.legend(loc="best")
    plt.xlabel("Pred")
    plt.ylabel("True")
    plt.show()
    plt.close()
    
    mse_tot = mean_squared_error(x_train_pred, y_train)
    print(f"mean squared error = {mse_tot:.7e}")

#     plt.title(r"$\Theta_{23}$")
    plt.xlabel("Train loss (MSE)")
    plt.ylabel("Events")
    plt.hist(mse, bins=50, density=False, alpha=0.8, color="tab:blue")
    plt.show()
    plt.close()

    return
```

First, we need to construct the “label kernel”, the input is the sin delta, and taking the arc sine of it; then we will get 2 degeneracy delta_cp. The green line is the result of the sine. We have known that the domain of arcsine is defined within -$\pi$ and $\pi$. Hence, there is one delta within this range and the other is out of it but still inside the -2$\pi$ to 2$\pi$. The red point represents the true label, in the training step it will be very close to the green sine curve, here we exaggerate it to easier to illustrate the labeling method. The predicted delta with orange color is inside the range of arcsine, I annotated it as 0; the other is blue which is outside the range of arcsine, labeled as 1. The criteria of which label number is it depending on the distance between the true label and the predicted label. We will choose the nearest as our predicted output label. After that, the output of the classification model will be the predicted label array and if we want to transform it back to $\delta_{cp}$, we can use the kernel constructed before.

This is the label kernel illustration:




```python
# this label kernel function is used to translate {1,0} array into delta

def translation_kernel(sin_pred, label_pred):
    print(f"The max, min of sine predictions: {sin_pred.max()}, {sin_pred.min()}")
    
    target = [0 if i<=0.5 else 1 for i in label_pred]
    delta = np.empty(len(sin_pred))
    theta1 = np.empty(len(sin_pred))
    theta2 = np.empty(len(sin_pred))
    for ind, ele in enumerate(sin_pred):
        # because the domain is differet and owing to the sine, it will differ with a negtive sign.
        theta = -np.arcsin(ele)
        if theta >= 0:
            theta1[ind], theta2[ind] = theta,  np.pi - theta
        else:
            theta1[ind], theta2[ind] = theta, -np.pi - theta
        if target[ind] == 0:
            delta[ind] = theta1[ind]
        else:
            delta[ind] = theta2[ind]
    return delta
```

```python
# Transpose the predictions

def predict_norm_transpose_delta(x_norm, y_norm):
    x = (x_norm + np.pi) * 180/np.pi
    y = (y_norm + np.pi) * 180/np.pi
    return x, y
def predict_norm_transpose_single_delta(x_norm):
    x = (x_norm + np.pi) * 180/np.pi
    return x
```

```python
from sklearn.metrics import mean_squared_error

# general analysis frame work
def analysis(x_train_pred, y_train) -> None:
    # set statistical parameters
    n = len(x_train_pred)
    dof = n - 2
    t = scipy.stats.t.ppf(0.975, dof)

    # linear fitting line
    slope, intercept = np.polyfit(x_train_pred, y_train, 1)
    slope, intercept, r, p, se = scipy.stats.linregress(x_train_pred, y_train)
    plt_min = np.min([np.min(x_train_pred), np.min(y_train)])
    plt_max = np.max([np.max(x_train_pred), np.max(y_train)])
    x_line = np.linspace(plt_min, plt_max, 100)
    y_line = np.polyval([slope, intercept], x_line)

    # acceleration kernel (to calculate mse, confident level, and prediction level)
    @njit(fastmath=True)
    def mse_kernel(pred, true, t_value=t, n=n, dof=dof, x_line=x_line, y_line=y_line):
        if true.shape != pred.shape:
            raise ValueError("True/Pred data should be the same shape!")
        mse = np.empty(len(true))
        for ind, (t, p) in enumerate(zip(true, pred)):
            mse[ind] = np.average(np.square(t - p))
        pred_mean = np.average(x_train_pred)
        sum_se = np.sum(np.square(true - pred))
        std_error = np.sqrt(1 / dof * sum_se)
        # confidence interval
        ci = (
            t_value
            * std_error
            * np.sqrt(
                1 / n
                + np.square(x_line - pred_mean)
                / np.sum(np.square(x_train_pred - pred_mean))
            )
        )
        # predicting interval
        pi = (
            t_value
            * std_error
            * np.sqrt(
                1
                + 1 / n
                + np.square(x_line - pred_mean)
                / np.sum(np.square(x_train_pred - pred_mean))
            )
        )
        return mse, ci, pi

    mse, ci, pi = mse_kernel(x_train_pred, y_train)

    fig = plt.figure(figsize=(7, 7), dpi=80)
    plt.scatter(x_train_pred, y_train, s=1, alpha=0.05, label="data")
    plt.plot(x_line, y_line, "--", lw=2, c="red", label="Fitting")
    plt.plot(
        [plt_min, plt_max], [plt_min, plt_max], "-", lw=1, c="orange", label="theo line"
    )
    plt.plot(
        x_line, y_line + pi, "--", lw=1, c="tab:red", label="95% prediction interval"
    )
    plt.plot(x_line, y_line - pi, "--", lw=1, c="tab:red")
    # plt.plot(x_line, y_line + ci, '--', c='tab:grey', label='95% confidence interval')
    # plt.plot(x_line, y_line - ci, '--', c='tab:grey')
    plt.xlim(plt_min, plt_max)
    plt.ylim(plt_min, plt_max)
    plt.legend(loc="best")
    plt.xlabel("Pred")
    plt.ylabel("True")
    plt.show()
    plt.close()
    
    mse_tot = mean_squared_error(x_train_pred, y_train)
    print(f"mean squared error = {mse_tot:.7e}")

    plt.xlabel("Train loss (MSE)")
    plt.ylabel("Events")
    plt.hist(mse, bins=50, density=False, alpha=0.8, color="tab:blue")
    plt.show()
    plt.close()

    return

# Transpose the predictions (Theta23)
def predict_norm_transpose_theta(x_norm, y_norm):
    x = x_norm * (MAX_Y - MIN_Y) + MIN_Y
    y = y_norm * (MAX_Y - MIN_Y) + MIN_Y
    return x, y
def predict_norm_transpose_single_theta(x_norm):
    x = x_norm * (MAX_Y - MIN_Y) + MIN_Y
    return x

# Translaiton kernel for delta classification
def translation_kernel(sin_pred, label_pred):
    print(f"The max, min of sine predictions: {sin_pred.max()}, {sin_pred.min()}")
    # force normalize to one
    for ind, ele in enumerate(sin_pred):
        if ele > 1:
            sin_pred[ind] = 1
        elif ele < -1:
            sin_pred[ind] = -1
    
    target = [0 if i<=0.5 else 1 for i in label_pred]
    delta = np.empty(len(sin_pred))
    theta1 = np.empty(len(sin_pred))
    theta2 = np.empty(len(sin_pred))
    for ind, ele in enumerate(sin_pred):
        # because the domain is differet and owing to the sine, it will differ with a negtive sign.
        theta = -np.arcsin(ele)
        if theta >= 0:
            theta1[ind], theta2[ind] = theta, np.pi - theta
        else:
            theta1[ind], theta2[ind] = theta, -np.pi - theta
        if target[ind] == 0:
            delta[ind] = theta1[ind]
        else:
            delta[ind] = theta2[ind]
    return delta

# Transpose the predictions (delta_cp)
def predict_norm_transpose_delta(x_norm, y_norm):
    x = (x_norm + np.pi) * 180/np.pi
    y = (y_norm + np.pi) * 180/np.pi
    return x, y
def predict_norm_transpose_single_delta(x_norm):
    x = (x_norm + np.pi) * 180/np.pi
    return x
```

```python
data_train = np.load(path + "neutrino_training_data.npz")  # ideal data
data_test = np.load(path + "neutrino_test_data.npz")  # pseudo-exp data

name_train = data_train.files
name_test = data_test.files

print(f"Train: {name_train}")
print(f"Test : {name_test}")

ve_train, vebar_train, vu_train, vubar_train, theta23_train, delta_train, ldm_train = map(lambda n: data_train[n], name_train)
ve_test, vebar_test, vu_test, vubar_test = map(lambda n: data_test[n], name_test)
```

```python
# create train and test data
X_train = np.stack((ve_train, vebar_train, vu_train, vubar_train), axis=-1)
Y_train = np.stack((theta23_train, delta_train, ldm_train), axis=-1)
X_test = np.stack((ve_test, vebar_test, vu_test, vubar_test), axis=-1)

print(
    f"""
The shape of x_train: {X_train.shape}
The shape of y_train: {Y_train.shape}
The shape of x_test : {X_test.shape}
"""
)
```

```python
from sklearn.linear_model import SGDClassifier
from sklearn import metrics

labeling = np.where(ldm_train > 0, 1, 0) # label NO or IO

ldm_class = SGDClassifier(max_iter=1000, tol=1e-3)
ldm_class.fit(X_train[:, :, 0], labeling)
train_predict = ldm_class.predict(X_train[:, :, 0])
test_predict = ldm_class.predict(X_test[:, :, 0])

fpr, tpr, thresholds = metrics.roc_curve(labeling, train_predict, pos_label=1)
plt.figure(figsize=(6, 6))
plt.plot(fpr, tpr, label="SDG model")
plt.plot([0, 1], [0, 1], label="Random level", ls="dashed", c="Grey")
plt.xlabel("False positive rate")
plt.ylabel("True positive rate")
plt.title(
    "Reciever operating characteristic, AUC = {:.2f}".format(metrics.auc(fpr, tpr))
)
plt.legend()
plt.tight_layout()
plt.show()
plt.close()

# clear unused variables
del ldm_class
```

```python
# check the test data is IO/NO

plt.hist(test_predict)
plt.title("LDM Test predict")
plt.ylabel("Events")
plt.xlabel("Label")
plt.show()
plt.close()
```

```python
"""
[Train X] normalize training data to [0,1]
"""
x_train_NH = X_train[ldm_train > 0]  # NO not IO
MAX_X = np.max(x_train_NH)
MIN_X = np.min(x_train_NH)
x_train_NH_norm = (x_train_NH - MIN_X) / (MAX_X - MIN_X)

"""
[Test X] normalize training data to [0,1]
"""
x_test_NH = X_test
# share the same MinMax values of X_train
x_test_NH_norm = (x_test_NH - MIN_X) / (MAX_X - MIN_X)

"""
[Train Y {Theta 23}] normalize training label to [0,1]
"""
# choose Theta_23 {0}
y_train_NH = Y_train[:, 0][ldm_train > 0]
MAX_Y = np.max(y_train_NH)
MIN_Y = np.min(y_train_NH)
y_train_NH_norm = (y_train_NH - MIN_Y) / (MAX_Y - MIN_Y)

"""
[Train Y {delta cp}] normalize training label
"""
# choose delta_cp {1}
y_train_NH = Y_train[:, 1][ldm_train > 0]
y_train_NH_norm_sin = np.sin(y_train_NH * np.pi / 180)  # degree -> rad [sin]
y_train_NH_norm_ang = np.pi * (y_train_NH - 180) / 180  # degree -> rad [-pi, pi]

print("Finish training labels...")

# clear unused variables
del (
    ve_train,
    vebar_train,
    vu_train,
    vubar_train,
    theta23_train,
    delta_train,
    ldm_train,
    ve_test,
    vebar_test,
    vu_test,
    vubar_test,
)
gc.collect()
```

```python
# create validation data
from sklearn.model_selection import train_test_split

# split the training dataset into training and validation, with test_size = 0.2, and random_state=10 with random_seed=2023
tf.random.set_seed(2023)
x_train, x_val, y_train, y_val = train_test_split(
    x_train_NH_norm, y_train_NH_norm, test_size=0.2, shuffle=True, 
    random_state=10
)
x_train_reg, x_val_reg, y_train_reg, y_val_reg = train_test_split(
    x_train_NH_norm, y_train_NH_norm_sin, test_size=0.2, shuffle=True, random_state=10
)
x_train_cla, x_val_cla, y_train_cla, y_val_cla = train_test_split(
    x_train_NH_norm, y_train_NH_norm_ang, test_size=0.2, shuffle=True, random_state=10
)
x_test = x_test_NH_norm

print(
    f"""
The shape of x_train     : {x_train.shape}
The shape of y_train     : {y_train.shape}
The shape of x_validation: {x_val.shape}
The shape of y_validation: {y_val.shape}
The shape of x_test      : {x_test.shape}
"""
)

del X_train, Y_train, X_test, x_train_NH_norm, y_train_NH_norm
```

```python
# load DCNN (with norm data) and do the analysis {Theta_23}

model = tf.keras.models.load_model("/kaggle/input/others/DCNN.h5", compile=False)
model.summary(expand_nested=True)
x_val_pred = model.predict(x_val)
theta_pred  = model.predict(x_test)
del model
```

```python
# Theta prediction checking

x, y = predict_norm_transpose_theta(x_val_pred[:, 0], y_val) 
analysis(x, y)
```

```python
# load reg and cla (with norm data) and do the analysis {delta_cp}
regression = tf.keras.models.load_model(
    "/kaggle/input/models/regression.h5", compile=False
)
classfication = tf.keras.models.load_model(
    "/kaggle/input/models/classification.h5", compile=False
)
regression.summary(expand_nested=True)
classfication.summary(expand_nested=True)

# validation checking
x_val_pred_sin = regression.predict(x_val_reg)
x_val_pred_ang = classfication.predict(x_val_cla)

# test data prediction
x_test_pred_sin = regression.predict(x_test)
x_test_pred_ang = classfication.predict(x_test)

del regression, classfication
```

```python
# sine delta prediction checking

analysis(x_val_pred_sin[:, 0], y_val_reg)
```

```python
# check delta classification checking

x_val_pred = translation_kernel(x_val_pred_sin[:, 0], x_val_pred_ang[:, 0])
x, y = predict_norm_transpose_delta(x_val_pred, y_val_cla)
analysis(x, y)

fig = plt.figure(figsize=(8,8), dpi=80)
plt.plot(x_val_pred[10:1500], 'r.', label="pred")
plt.plot(y_val_cla[10:1500], 'b.', label="true")
plt.legend(loc="best")
plt.ylabel("degrees")
plt.xlabel("Event number")
plt.show()
```

```python
# label kernel translation

delta_pred = translation_kernel(x_test_pred_sin[:,0], x_test_pred_ang[:,0])
```

```python
# original (raw) domain mapping

x_theta_dump = predict_norm_transpose_single_theta(theta_pred)
x_delta_dump = predict_norm_transpose_single_delta(delta_pred)

print(x_theta_dump.shape)
print(x_delta_dump.shape)
print(x_theta_dump.max(), x_theta_dump.min())
print(x_delta_dump.max(), x_delta_dump.min())
```

```python
dump = np.concatenate((x_delta_dump, x_theta_dump[:,0]))
print(dump.shape)
```
